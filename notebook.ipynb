# Executive Summary: Employee Compensation Analysis

### 1. Problem Statement
Analyze departmental pay disparities and regional compensation trends across employee roles.

### 2. Data Overview
Dataset includes 8 employee records featuring ID, Name, Department, Salary (INR), and City[cite: 1].

### 3. Key Insights
- **Departmental Disparity:** Finance holds the highest average salary (90,000 INR), whereas Sales displays the lowest average pay[cite: 1].
- **Geographic Variation:** Staff in Bangalore and Mumbai command higher compensation tiers relative to Chennai[cite: 1].
- **Distribution Range:** The majority of employee salaries fall between 40,000 and 70,000 INR[cite: 1].

### 4. Recommendations
- Re-align base salary bands for Sales roles to meet market standard baselines.
- Standardize cost-of-living allowances across regional office locations.

### 5. Limitations
The small sample size ($N=8$) limits statistical generalizability across larger organizations[cite: 1].

```python
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Ensure output folder exists
os.makedirs("reports/figures", exist_ok=True)

# Global styling configuration
sns.set_theme(style="whitegrid")
PRIMARY_COLOR = "#2b5c8f"

# Load dataset
df = pd.read_csv("employees.csv")

# 1. Bar Chart
plt.figure(figsize=(8, 5))
avg_salary = df.groupby("department")["salary"].mean().reset_index()
sns.barplot(data=avg_salary, x="department", y="salary", color=PRIMARY_COLOR)
plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary (INR)")
plt.tight_layout()
plt.savefig("reports/figures/01_bar_chart.png")
plt.show()

# 2. Histogram
plt.figure(figsize=(8, 5))
sns.histplot(df["salary"], bins=5, kde=True, color=PRIMARY_COLOR)
plt.title("Salary Distribution")
plt.xlabel("Salary (INR)")
plt.ylabel("Count")
plt.tight_layout()
plt.savefig("reports/figures/02_histogram.png")
plt.show()

# 3. Scatter Plot
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="city", y="salary", hue="department", s=100)
plt.title("Salary vs City by Department")
plt.xlabel("City")
plt.ylabel("Salary (INR)")
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig("reports/figures/03_scatter_plot.png")
plt.show()

# 4. Line Chart
plt.figure(figsize=(8, 5))
sns.lineplot(data=df, x="id", y="salary", marker="o", color=PRIMARY_COLOR)
plt.title("Salary Trend Across Employee IDs")
plt.xlabel("Employee ID")
plt.ylabel("Salary (INR)")
plt.tight_layout()
plt.savefig("reports/figures/04_line_chart.png")
plt.show()

# 5. Box Plot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="department", y="salary", palette="Blues")
plt.title("Salary Spread by Department")
plt.xlabel("Department")
plt.ylabel("Salary (INR)")
plt.tight_layout()
plt.savefig("reports/figures/05_box_plot.png")
plt.show()
```
